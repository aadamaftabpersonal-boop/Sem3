Ch1:
How do we build applications that keep working correctly, remain fast as they grow, and are still understandable/changeable years later?

reliable + scalable + manageable

1. the big picture
modern systems are data-intensive not compute-intensive
compute-intensive:
i have a difficult calculation to perform

for example:
input->cpu->huge computation->output

examples:
rendering a 3d scene
training a neural network
solving a mathematical optimization problem
cryptographic computation

bottleneck is usually cpu/gpu power

data-intensive
instead you have:
huge amounts of data
lots of users
lots of requests
rapidly changing data
requirements to store/search/process that data

for example, imagine instagram
a user uploads a photo

your system may need to:
Store photo
    ↓
Store metadata
    ↓
Update user's feed
    ↓
Update followers' feeds
    ↓
Generate notifications
    ↓
Update search index
    ↓
Update recommendation system
    ↓
Send analytics events


the challenge isnt:
can my cpu calculate this?

it is:
"How do I move, store, process, synchronize, and retrieve all this data reliably at huge scale?"


the standard building blocks of data systems

databases:
store data so that you can retrieve it later
example:
user:
id = 42
name = aadam
age = 19

database gives you persistence

if your server crashes:
RAM - no
database correct
the data survives

examples:
postgresql
mysql
mongodb
cassandra
