OS can be understood by 3 fundamental principles:
1. virtualisation
2. concurrency
3. persistence

applications <-> OS <-> hardware

The hardware is complicated and limited. The OS makes it appear simpler, safer, and more powerful to applications.

Virtualisation:
suppose u have:
8gb ram, 1 cpu, 1ssd
but you run:
chrome, vscode, spotify, discord, a game, c++ program

each program behaves as if it has resources available to itself

but there is only one cpu
the OS creates an illusion that each application has its own cpu
thats virtualization

Physical CPU
     ↓
     OS
     ↓
 ┌────┬────┬────┬────┐
 ↓    ↓    ↓    ↓
App1 App2 App3 App4

makes each application feel like it has its own cpu

The OS rapidly switches the CPU between programs.

This leads directly to one of the major OS topics:

CPU virtualization → Processes

memory virtualization:
suppose you have 8gb ram
you run two programs

program A might think
Address 0 → my data
Address 1000 → my data

program B might think:
Address 0 → my data
Address 1000 → my data

how can both use same addresses?
Because the OS + hardware provide virtual memory.

             Virtual memory
             ↓          ↓
        Program A    Program B
             ↓          ↓
             OS
              ↓
       Physical RAM

OS translates each programs virtual addresses into physical memory locations

Take a physical resource and create an abstraction that makes it look like each program has its own resource.

Concurrency:
say an application has to do many things at once
say, web browser - downloads file, renders a webpage, plays a video, handles mouse input all at once

Concurrency doesn't necessarily mean things literally execute simultaneously.

consider single CPU:
A A A B B B A A C C C B B

the cpu rapidly switches between activities

to you, they appear to be happening simultaneously

with multiple cpu cores, some activities can genuinely execute simultaneously
Core 1 → A A A A
Core 2 → B B B B
Core 3 → C C C C

say cpu has 2 threads - thread A and thread B
say a program
counter=10;
counter++;
counter++;
expected output = 12

but program may give 11 as on swtiching context between 2 threads the second thread may remember old value and not updated value
Two threads are using shared data without properly coordinating their accesses. - problem




this is called race around condition

Persistence
say you create a notes.txt file, you shutdown and restart. you expect notes.txt to be present

   ↓
RAM contains data
   ↓

   ↓

storage on SSDs/HDDs retain info:
Power ON
   ↓
SSD contains data
   ↓
Power OFF
   ↓
Data remains

The OS provides abstractions that make persistent storage easier to use.
