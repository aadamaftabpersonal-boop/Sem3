signal is a quantity that changes with one or more independent variables
ex: temperature changes with hour --> T = f(t)

every signal has dependent and independent variables
ex: speech with time s(t)

continuous signal doesnt mean continuous graph
a continuous signal exists every instant of time x(t)

graph has no gaps, not broken

discrete graph - signals only exist at integer points of x
x[n] = x(nT) T is the sampling period which need not be integer

deterministic vs stochastic:
deterministic - at a given time, the value of signal is known
ex: x=6cos(5t)

stochastic - at a given time, the value of signal is unknown
ex: x=6cost(5t) + noise

sampling:
capturing signal at finite time intervals
smaller t = more sample -> more accurate -> fps

continuous periodic time signal => x(t) = x(t+T)
discrete periodic time signal => x[n] = x[n+N], where N is an integer time period

addition of a periodic and non periodic signal is non periodic

y(t) = x(bt)
0<b<1 -> expansion
b>1 -> compression

time reversal -> folding:
replace t with -t

x(t-a) -> graph shape remains same but it shifts a units forward
