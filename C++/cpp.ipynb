Any person can register as a user with a username she choses. If a username already exists, registration fails. A registered user can login to the app. After login, an user can create a post by uploading one or more photos and specifying a caption for it. A photo is basically a 2D matrix of pixel values, a recorded timestamp, and the source (e,g, camera, WhatsApp downloads etc). User can also delete a post in the following manner: she scrolls through her posts and stops at the intended one that she wants to delete, and then deletes it.
